```python
import os

# ==========================
# CONFIGURACIÓN
# ==========================

CARPETA = "mis_cuentos"

INICIO = 206
FIN = 1000

os.makedirs(CARPETA, exist_ok=True)

# ==========================
# CREAR TXT VACÍOS
# ==========================

for i in range(INICIO, FIN + 1):

    nombre_archivo = f"cuento_{i:03d}.txt"
    ruta = os.path.join(CARPETA, nombre_archivo)

    if os.path.exists(ruta):
        print(f"Ya existe: {nombre_archivo}")
        continue

    contenido = f"""TITULO:

FUENTE:

AUTOR:

TEMATICA: princesas_castillos

IDIOMA: español

==================================================

PEGA AQUÍ EL CUENTO

"""

    with open(ruta, "w", encoding="utf-8") as f:
        f.write(contenido)

print(f"Se generaron los archivos desde cuento_{INICIO:03d}.txt hasta cuento_{FIN:03d}.txt")
```
